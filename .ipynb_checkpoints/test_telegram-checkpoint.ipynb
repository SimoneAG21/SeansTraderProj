# Test Telegram Connection for TradingV1
Verify connection to Telegram using ConfigManager and telethon.

```python
import sys
import os
import asyncio

# Add project root to path
project_root = os.path.abspath(os.path.join(os.getcwd(), '.'))
sys.path.append(project_root)

from telethon import TelegramClient
from helper.config_manager import ConfigManager
from helper.Logger import Logging

# Initialize config and logger
config = ConfigManager(config_file='config/channel_sync_config.yaml')
logger = Logging(config=config, instance_id='test_telegram')
logger.set_log_file('test_telegram')

# Get Telegram config
tele_config = config.get_section('telegram')
print(f'Telegram config: {tele_config}')

# Test Telegram connection
async def test_connection():
    client = TelegramClient(
        tele_config['session_path'],
        tele_config['api_id'],
        tele_config['api_hash']
    )
    try:
        logger.info('Attempting to connect to Telegram...')
        await client.connect()
        if not await client.is_user_authorized():
            logger.info('Session not authorized; starting login...')
            await client.start(phone=tele_config['phone_number'])
        else:
            logger.info('Using existing authorized session')
        logger.info('Successfully connected to Telegram!')
        # Test fetching dialogs
        dialogs = await client.get_dialogs(limit=5)
        print(f'Fetched {len(dialogs)} dialogs: {[d.name for d in dialogs]}')
    except Exception as e:
        logger.error(f'Failed to connect to Telegram: {e}')
        raise
    finally:
        if client.is_connected():
            await client.disconnect()
            logger.info('Disconnected from Telegram')
        logger.close_log()

# Run async test
await asyncio.run(test_connection())
```
