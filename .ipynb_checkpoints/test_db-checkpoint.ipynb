# Test Postgres Connection for TradingV1
Verify connection to postgresql-clean-45989 using ConfigManager and sqlalchemy.

```python
import sys
import os

# Add project root to path
project_root = os.path.abspath(os.path.join(os.getcwd(), '.'))
sys.path.append(project_root)

from sqlalchemy import create_engine
from sqlalchemy.sql import text
from helper.config_manager import ConfigManager
from helper.Logger import Logging

# Initialize config and logger
config = ConfigManager(config_file='config/combined_config.yaml')
logger = Logging(config=config, instance_id='test_db')
logger.set_log_file('test_db')

# Get database config
db_config = config.get_section('database')
print(f'Database config: {db_config}')

# Create connection string
connection_string = f"postgresql+psycopg2://{db_config['user']}:{db_config['password']}@{db_config['host']}:{db_config['port']}/{db_config['database']}"
print(f'Connection string: {connection_string}')

# Test connection
try:
    engine = create_engine(connection_string)
    with engine.connect() as conn:
        logger.info('Successfully connected to Postgres!')
        # Test a simple query
        result = conn.execute(text('SELECT 1'))
        print('Query result:', result.fetchone())
except Exception as e:
    logger.error(f'Failed to connect to Postgres: {e}')
    raise
finally:
    logger.close_log()
```
